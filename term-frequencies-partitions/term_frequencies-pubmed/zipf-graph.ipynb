```python
import pandas as pd
import matplotlib.pyplot as plt

# Load PubMed noun frequencies
df = pd.read_csv("noun_frequencies_pubmed.csv")

# Sort terms from highest to lowest frequency
df = df.sort_values("frequency", ascending=False).reset_index(drop=True)

# Rank starts at 1
df["rank"] = df.index + 1

# Plot
plt.figure(figsize=(16, 9))

plt.plot(
    df["rank"],
    df["frequency"]
)

plt.xscale("log")
plt.yscale("log")

plt.xlabel("Rank", fontsize=16)
plt.ylabel("Frequency", fontsize=16)
plt.title("Complete Frequency Distribution (PubMed)", fontsize=20)

plt.tick_params(axis="both", labelsize=14)

plt.tight_layout()
plt.show()
```
