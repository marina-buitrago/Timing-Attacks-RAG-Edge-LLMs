```python
import pandas as pd
import matplotlib.pyplot as plt

# Cargar datos
df = pd.read_csv("noun_frequencies_pubmed.csv")

# Ordenar por frecuencia (por si no lo estuviera)
df = df.sort_values("frequency", ascending=False).reset_index(drop=True)

# Crear la figura
plt.figure(figsize=(12, 5))
plt.plot(df.index, df["frequency"], linewidth=1)

plt.xlabel("Ranking del término")
plt.ylabel("Frecuencia")
plt.title("Distribución de frecuencias de los términos")
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```
