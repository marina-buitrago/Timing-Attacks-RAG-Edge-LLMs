# PubMed white-box topic inference, final: query-budget reduction

Runs the final (leak-free, per-noun, 500-tree) topic experiment for smol, qwen and granite with 100%, 75%, 50%, 25%, 10% and 5% of the queries. Every noun keeps the same share of its queries, split evenly across short, medium and long (same selection as the original budget notebooks).

* `random` templates: each noun keeps its own random templates (seed 42).
* `shared` templates: every noun keeps the same template indices (seed 42).

Gain over random = (Macro-F1 − 1/5) / (1/5).

Results: `pubmed_white_box_topic_budget_results/` with `budget_summary.csv` and the budget curves for both template rules.

```python
import sys
from pathlib import Path

# final_topic.py lives in classificators/, two levels up from this notebook.
sys.path.insert(0, str(Path.cwd().parents[1]))
import final_topic as rt
import pandas as pd

BOX = "white"
CACHE = False
DATASET = "pubmed"
ROOT = rt.budget_root(BOX, CACHE, DATASET)
ROOT.mkdir(parents=True, exist_ok=True)

rows = pd.concat(
    [rt.run_budget_model(BOX, CACHE, DATASET, model, ROOT) for model in rt.MODELS],
    ignore_index=True,
)
budget_summary = rt.write_budget_summary(rows, ROOT)
rt.plot_all_budget_curves(BOX, CACHE, DATASET, ROOT, budget_summary)
budget_summary[["template_rule", "model", "budget_pct", "macro_f1", "gain_over_random_pct", "accuracy"]].round(3)
```
